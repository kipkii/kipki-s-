# 분위수 (quartile)



분위수는 데이터를 순서대로 늘어 놓았을 떄 이를 균등한 간격으로 나누는 기준점 입니다. 예를 들어 이분위수는 전체데이터를 두구간으로 나누기 때문에 중앙값에 해당
따라서 사분위수는 3개가 나오고, 각각 25, 50 75에 해당
25에 대항하는 값을 1사분위수, 중앙값을 2 사분위수, 75값을 3사분위 수라고도 부릅니다.

```python
ns_book6['대출건수'].quantile(0.25)
```

```python
ns_book6['대출건수'].quantile([0.25,0.5,0.75])
```

```python
pd.Series([1,2,3,4,5]).quantile(0.9)
```

```python
pd.Series([1,2,3,4,5]).quantile(0.9, interpolation='midpoint')
```

```python
pd.Series([1,2,3,4,5]).quantile(0.9, interpolation='nearest')
```

두지점사이에 놓인 특정 위치 값을 구하는 방법을 보간(interploation)이라고 부른다
interpolation매개변수는 linear외에도 여러가지 다른 보간 방식을 제공
midpoint는 분위수에 상관없이 무조건 두 수 사이의 중앙값을 사용한다.
nearest는 두수중에서 가까운 값을 사용한다.
